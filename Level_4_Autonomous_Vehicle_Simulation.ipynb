```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time = np.arange(0, 10, 0.1)

initial_speed = 20       # m/s
vehicle_speed = []
vehicle_distance = []

speed = initial_speed
distance = 0

for t in time:

    # Pedestrian detected after 4 seconds
    if t >= 4:
        speed = max(0, speed - 4 * 0.1)

    distance += speed * 0.1

    vehicle_speed.append(speed)
    vehicle_distance.append(distance)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(time, vehicle_speed)

plt.xlabel("Time (seconds)")
plt.ylabel("Vehicle Speed (m/s)")
plt.title("Level 4 Autonomous Vehicle - Automatic Braking")
plt.grid(True)

plt.show()
```

```python
pedestrian_detected = False
vehicle_action = []

for t in time:

    if t >= 4:
        pedestrian_detected = True

    if pedestrian_detected:
        action = "STOP"
    else:
        action = "CONTINUE"

    vehicle_action.append(action)

print("Before pedestrian detection:", vehicle_action[20])
print("After pedestrian detection:", vehicle_action[50])
```

```python
plt.figure(figsize=(8, 5))

plt.plot(time, vehicle_speed)

plt.axvline(x=4, linestyle="--", label="Pedestrian Detected")

plt.xlabel("Time (seconds)")
plt.ylabel("Vehicle Speed (m/s)")
plt.title("Level 4 Autonomous Vehicle - Automatic Braking")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(time, vehicle_distance)

plt.axvline(x=4, linestyle="--", label="Pedestrian Detected")

plt.xlabel("Time (seconds)")
plt.ylabel("Vehicle Distance (m)")
plt.title("Vehicle Distance vs Time")
plt.legend()
plt.grid(True)

plt.show()
```

```python
print("Autonomous Vehicle Decision Log")
print("--------------------------------")

for i in range(0, len(time), 10):
    if time[i] < 4:
        print(f"Time {time[i]:.1f}s → CONTINUE")
    else:
        print(f"Time {time[i]:.1f}s → STOP")
```

```python
pedestrian_distance = []

for t in time:
    if t < 4:
        distance_to_pedestrian = 50 - (t * 5)
    else:
        distance_to_pedestrian = max(0, 30 - ((t - 4) * 2))

    pedestrian_distance.append(distance_to_pedestrian)

plt.figure(figsize=(8, 5))

plt.plot(time, pedestrian_distance)

plt.axvline(x=4, linestyle="--", label="Pedestrian Detected")

plt.xlabel("Time (seconds)")
plt.ylabel("Distance to Pedestrian (m)")
plt.title("Pedestrian Distance Detection")
plt.legend()
plt.grid(True)

plt.show()
```
